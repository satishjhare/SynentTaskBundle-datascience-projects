```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Iris dataset
df = pd.read_csv("/content/drive/MyDrive/Datascience_Project /Iris.csv")

# Display first rows
print(df.head())
```

```python
# Bar Chart
sns.countplot(x='Species', data=df)
plt.title("Species Count")
plt.show()
```

```python
# Histogram
plt.hist(df['SepalLengthCm'], bins=20)
plt.title("Sepal Length Distribution")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```

```python
# Scatter Plot
sns.scatterplot(
    x='SepalLengthCm',
    y='PetalLengthCm',
    hue='Species',
    data=df
)
```

```python
plt.title("Sepal Length vs Petal Length")
plt.show()
```
